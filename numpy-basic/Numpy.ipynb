```python
import numpy as np
A = np.array([
    [2, 3],
    [4, 5]
])

B = np.array([
    [1, 2],
    [3, 4]
])
```

```python
# addition, multiplication and division
print ("Addition" , A + B)
print ("Multiplication", A * B)
print ("Division", A / B)
print ("Scalar Multiplication", 2 * B)
print ("Scalar Division",  B / 2)
```

```python
print(np.dot(A , B)) # return a scalar  (3 * 1) + (4 * 2)
```

```python
print(np.linalg.norm(A)) # return a scalar root(3 ^ 2 * 4 ^ 2)
```

```python
print("Element wise multiplication\n", A * B) # Element wise multiplication
print("rows × columns\n", A @ B) # rows x column
print(A.shape)
```

```python
# vector and numpy array -> vector (we will have [age, salary] @ Weights W = [[]]).
# Weights : How much should each feature contribute to the prediction? from training ML model generates weights
# shape -> (2 x 3)(3 x 1)=(2 x 1)
#Weight influece the feature

x = np.array([10, 20])
print(x @ A)
```

```python
# bias uplift the o/p
# y = W + b
b = np.array([5,10])
print("O/P with bias : " , (x @ A) + b)
```
